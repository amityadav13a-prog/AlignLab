```python
#Formatting function
# ### Instruction:
# Translate this sentence.
# ### Input:
# Hello world.
# ### Response:
# Hola mundo.
```

```python
def format_example(row):
    instruction = row["instruction"].strip()
    input_text = row["input"].strip()
    output = row["output"].strip()

    if input_text:
        prompt = f"""### Instruction:
{instruction}

### Input:
{input_text}

### Response:
{output}"""
    else:
        prompt = f"""### Instruction:
{instruction}

### Response:
{output}"""

    return prompt

# nya col. bna k store krega
df["formatted_text"] = df.apply(format_example, axis=1)
# printing new col
print(df["formatted_text"].iloc[0])
```

```python
# response-only masking ke liye hume prompt aur response ki boundary exactly pata honi chahiye.
# Isliye train aur validation ke liye ye function banayenge:
def prepare_sft_example(row):
    instruction = row["instruction"].strip()
    input_text = row["input"].strip()
    output = row["output"].strip()

    if input_text:
        prompt = (
            f"### Instruction:\n"
            f"{instruction}\n\n"
            f"### Input:\n"
            f"{input_text}\n\n"
            f"### Response:\n"
        )
    else:
        prompt = (
            f"### Instruction:\n"
            f"{instruction}\n\n"
            f"### Response:\n"
        )

    full_text = prompt + output

    return {
        "prompt": prompt,
        "response": output,
        "text": full_text
    }
```

```python
#Function apply
train_prepared = train_df.apply(
    prepare_sft_example,
    axis=1,
    result_type="expand"
)

val_prepared = val_df.apply(
    prepare_sft_example,
    axis=1,
    result_type="expand"
)

print(train_prepared.columns.tolist())
print(val_prepared.columns.tolist())
```

```python
#Labels
#Full tokenization + masking function
def tokenize_and_mask(row):
    prompt = row["prompt"]
    text = row["text"]

    full_tokens = tokenizer(
        text,
        truncation=True,
        max_length=MAX_LENGTH,
        padding=False
    )

    prompt_tokens = tokenizer(
        prompt,
        add_special_tokens=False,
        truncation=False
    )

    input_ids = full_tokens["input_ids"]
    attention_mask = full_tokens["attention_mask"]

    prompt_length = len(prompt_tokens["input_ids"])

    # If truncation removed the entire response,
    # this example provides no useful response loss.
    if prompt_length >= len(input_ids):
        return {
            "input_ids": input_ids,
            "attention_mask": attention_mask,
            "labels": [-100] * len(input_ids)
        }

    labels = [-100] * len(input_ids)

    for i in range(prompt_length, len(input_ids)):
        labels[i] = input_ids[i]

    return {
        "input_ids": input_ids,
        "attention_mask": attention_mask,
        "labels": labels
    }
```

```python
#single example testing 
test_prepared = train_prepared.iloc[0]

test_result = tokenize_and_mask(test_prepared)

print("Input IDs:")
print(test_result["input_ids"])

print("\nLabels:")
print(test_result["labels"])

print("\nAttention mask:")
print(test_result["attention_mask"])
```

```python
#Masking visually verify
for token_id, label in zip(
    test_result["input_ids"],
    test_result["labels"]
):
    token = tokenizer.decode([token_id])

    print(
        repr(token),
        "→",
        label
    )
```

```python
# whole train dataset ko tokenize
train_tokenized = train_prepared.apply(
    tokenize_and_mask,
    axis=1,
    result_type="expand"
)

val_tokenized = val_prepared.apply(
    tokenize_and_mask,
    axis=1,
    result_type="expand"
)

print("Train:", train_tokenized.shape)
print("Validation:", val_tokenized.shape)
print("\nColumns:")
print(train_tokenized.columns.tolist())
```
